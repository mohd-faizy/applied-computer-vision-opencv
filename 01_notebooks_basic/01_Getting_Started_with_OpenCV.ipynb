## Part 1: The Top 5 Core Image Concepts

Before writing a single line of code, you must understand how a computer "sees" an image. To a machine, a photo is not a graphic; it is raw data.

**1. What is an Image?**
- An image is a **2D grid** (or a matrix) made up of rows and columns.
- If an image is 100 pixels wide and 100 pixels tall, it is a matrix of 10,000 tiny dots ($100 \times 100$).

**2. Pixels & The 0-255 Range**
- A pixel is the smallest individual unit (a single dot) of an image.
- Each pixel holds numerical data that dictates its color.
- These values range from **0 to 255**.
- **0** means the color is completely turned off (Black).
- **255** means the color is at its absolute maximum intensity.

**3. Width and Height**
- **Width (x-axis):** The total number of pixels horizontally (Left to Right / Columns).
- **Height (y-axis):** The total number of pixels vertically (Top to Bottom / Rows).

**4. Color Channels (RGB vs. BGR)**
A color channel stores one specific part of the color information for each pixel.
- **Grayscale Image:** Has **1 channel**. A single number from 0-255 dictates how bright or dark that specific pixel is.
- **Colored Image:** Has **3 channels**. While the world generally uses **RGB**, OpenCV officially uses **BGR** (Blue, Green, Red) due to legacy camera hardware standards.

**5. Image Formats (Encodings)**
| Format | Description |
|---|---|
| `.jpg` | Lossy compression, small file size. |
| `.png` | Lossless, supports transparency (alpha channel). |
| `.bmp` | Uncompressed, raw pixels. Large size. |
| `.tiff` | Super high resolution, used in medical/satellite imaging. |


## Load Sample Image

When you read an image, you are loading its data (pixels) into your program's memory.

> **Tip**  
> Always verify that your image loaded successfully. `cv2.imread()` returns `None` if the path is incorrect, rather than throwing a Python exception!


```python
image_path = "../data/images/lena.jpg"
image = cv2.imread(image_path)

if image is None:
    print(f"Error: Image not found at {image_path}! Check your file path.")
else:
    print("Image loaded successfully.")
```

## Displaying an Image

To view the matrix visually, you need a combination of three specific functions.

| Function | Purpose |
|---|---|
| `cv2.imshow` | Opens a window and shows the image |
| `cv2.waitKey` | Pauses the code so the window doesn't close instantly |
| `cv2.destroyAllWindows` | Cleans up memory and closes windows |


```python
if image is not None:
    # 1. Opens a window and shows the image
    cv2.imshow("Original Image", image)
    
    # 2. Pauses the code (0 means wait indefinitely for a key press)
    cv2.waitKey(0)
    
    # 3. Cleans up memory and closes the window safely.
    cv2.destroyAllWindows()
```

## Saving an Image

Once you are done manipulating an image in memory, you write it back to your local disk using `cv2.imwrite`.


```python
if image is not None:
    output_path = "output_lena.png"
    success = cv2.imwrite(output_path, image)
    
    if success:
        print(f"Image saved successfully as {output_path}")
    else:
        print("Failed to save the image.")
```

## Checking Image Dimensions

`.shape` is a NumPy attribute that tells you the exact dimensions and channels of your loaded matrix.

> **Note**  
> If the image is Grayscale, `.shape` returns only `(Height, Width)`. For color images, it returns `(Height, Width, Channels)`.


```python
if image is not None:
    # Returns a tuple: (Height, Width, Channels)
    dimensions = image.shape
    height, width, channels = dimensions
    
    print(f"Image Dimensions: {dimensions}")
    print(f"Height: {height} pixels")
    print(f"Width: {width} pixels")
    print(f"Channels: {channels} (BGR)")
```

## Grayscale Conversion

Converting an image to grayscale reduces the processing time and complexity since the computer only has to track 1 channel instead of 3. This is essential for advanced tasks like face or object detection.


```python
if image is not None:
    # Converts from default BGR to Grayscale
    gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    
    cv2.imshow("Grayscale Image", gray_image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
    
    # Notice how the shape has changed
    print(f"Grayscale Dimensions: {gray_image.shape}")
```

## Common Mistakes
> **Warning**  
> - Forgetting `cv2.waitKey(0)` will cause the image window to instantly crash or freeze.
> - Assuming `cv2.imread()` throws an error on a bad path. It simply returns `None`, which causes a `NoneType` error later on when you try to display or process it.


## Key Takeaways
- OpenCV reads images in BGR format, not RGB.
- Images are just NumPy matrices of shape `(Height, Width, Channels)`.
- The intensity range is 0 to 255.
