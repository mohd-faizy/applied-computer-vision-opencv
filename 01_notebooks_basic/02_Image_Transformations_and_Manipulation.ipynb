## Load Sample Image

We will load a sample image to demonstrate transformations.


```python
image_path = "../data/images/lena.jpg"
image = cv2.imread(image_path)

if image is None:
    print(f"Error: Image not found at {image_path}! Check your file path.")
else:
    print("Image loaded successfully.")
```

## Resizing & Scaling Images (`cv2.resize`)

**Why resize an image?**
- **Faster Processing:** Smaller images process much faster in AI algorithms.
- **Standardization:** ML models require standardized input sizes.
- **Display:** Fitting large images onto smaller screens.

> **Warning**  
> **The Golden Rule of OpenCV Resizing:** OpenCV's `resize` function explicitly requires dimensions to be passed as `(Width, Height)`. This is the opposite of the `(Height, Width)` returned by `.shape`. Passing them backward will distort your image!


```python
if image is not None:
    # Resize it to exactly 300 pixels wide and 300 pixels tall
    resized_image = cv2.resize(image, (300, 300))
    
    cv2.imshow("Original", image)
    cv2.imshow("Resized (300x300)", resized_image)
    
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

## Cropping Using NumPy Slicing

Cropping is cutting out a specific rectangular region of interest (ROI) from a larger image. Because an image is a NumPy array, we don't use an OpenCV function—we use raw Python matrix slicing.

**The Slicing Syntax:** `image[start_y : end_y, start_x : end_x]`
- You **MUST** provide the Y (Height) coordinates first, followed by the X (Width) coordinates.


```python
if image is not None:
    # Let's want a box where:
    # Y goes from 100 to 200 (100 pixels tall)
    # X goes from 50 to 150 (100 pixels wide)
    
    cropped_image = image[100:200, 50:150]
    
    cv2.imshow("Original", image)
    cv2.imshow("Cropped Part", cropped_image)
    
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

## Image Rotation (`cv2.getRotationMatrix2D` & `cv2.warpAffine`)

Spinning an image around a specific pivot point. 

- **Step 1:** Find the exact center using integer division `//`.
- **Step 2:** Create the rotation matrix using `cv2.getRotationMatrix2D(center, angle, scale)`.
  - *Angle:* Positive = Counter-Clockwise. Negative = Clockwise.
- **Step 3:** Apply the matrix using `cv2.warpAffine`.


```python
if image is not None:
    # 1. Grab Height and Width from the shape attribute
    height, width = image.shape[:2]
    
    # 2. Calculate the exact center coordinates (x, y)
    center = (width // 2, height // 2)
    
    # 3. Create the transformation matrix (90 degrees counter-clockwise)
    rotation_matrix = cv2.getRotationMatrix2D(center, 90, 1.0)
    
    # 4. Apply the rotation
    rotated_image = cv2.warpAffine(image, rotation_matrix, (width, height))
    
    cv2.imshow("Rotated 90 Degrees", rotated_image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

## Image Flipping (`cv2.flip`)

Mirroring the image data across an axis. 

**The Flip Codes:**
- `1`: Horizontal Flip (Mirror effect).
- `0`: Vertical Flip (Upside down).
- `-1`: Both Horizontal and Vertical Flip combined.


```python
if image is not None:
    flipped_horizontal = cv2.flip(image, 1)
    flipped_vertical = cv2.flip(image, 0)
    flipped_both = cv2.flip(image, -1)
    
    cv2.imshow("Flipped Horizontally", flipped_horizontal)
    cv2.imshow("Flipped Vertically", flipped_vertical)
    cv2.imshow("Flipped Both", flipped_both)
    
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

## Common Mistakes
> **Warning**  
> - Mixing up `(Width, Height)` in OpenCV functions and `(Height, Width)` in NumPy slicing.
> - Using floating-point numbers instead of integer division `//` for coordinates. OpenCV expects integer pixel coordinates.


## Key Takeaways
- Resizing relies on `cv2.resize()`.
- Cropping relies purely on NumPy array slicing.
- Complex geometric transformations like rotation require computing a matrix first, then applying it via `cv2.warpAffine()`.
