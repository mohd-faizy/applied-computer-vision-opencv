## Load Sample Image

Instead of an external image, let's create a blank black canvas using NumPy to practice drawing.
A black canvas is simply an array filled with zeros.


```python
# Create a black image: 500x500 pixels, 3 color channels (BGR), data type uint8
canvas = np.zeros((500, 500, 3), dtype="uint8")

cv2.imshow("Blank Canvas", canvas)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

## The OpenCV Coordinate System

Before drawing, you must understand the canvas. OpenCV treats your image like a graph:
- **The Origin `(0, 0)`** is at the **Top-Left** corner.
- **X-axis** moves horizontally left to right.
- **Y-axis** moves vertically top to bottom.
- Coordinates are always passed as tuples: `(X, Y)`.


## Drawing Lines (`cv2.line`)

Lines are drawn by specifying a starting point (`pt1`) and an ending point (`pt2`).
- **Syntax:** `cv2.line(image, pt1, pt2, color, thickness)`


```python
# Define Start (pt1) and End (pt2) points -> (X, Y)
pt1 = (50, 100)   # 50 pixels right, 100 pixels down
pt2 = (450, 100)  # 450 pixels right, 100 pixels down

# Define Color (BGR) and Thickness
color = (255, 0, 0)  # Pure Blue
thickness = 4

cv2.line(canvas, pt1, pt2, color, thickness)

cv2.imshow("Line Drawing", canvas)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

## Drawing Rectangles (`cv2.rectangle`)

You only need two opposite diagonal corners: the **Top-Left** and the **Bottom-Right**.
- **Syntax:** `cv2.rectangle(image, top_left, bottom_right, color, thickness)`

> **Tip**  
> If you want a SOLID, filled-in rectangle, set the thickness to `-1`.


```python
# Top-Left corner (X1, Y1)
top_left = (50, 150)

# Bottom-Right corner (X2, Y2)
bottom_right = (200, 300) 

# Draw a red rectangle (Thickness = 3)
cv2.rectangle(canvas, top_left, bottom_right, (0, 0, 255), 3)

# Draw a filled green rectangle next to it
cv2.rectangle(canvas, (250, 150), (400, 300), (0, 255, 0), -1)

cv2.imshow("Rectangle Drawing", canvas)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

## Drawing Circles (`cv2.circle`)

A circle requires a **Center Point** and a **Radius** (distance from the center to the edge in pixels).
- **Syntax:** `cv2.circle(image, center, radius, color, thickness)`


```python
# Define the center coordinate (X, Y)
center = (250, 400)

# Define the radius in pixels
radius = 50

# Draw a filled blue circle
cv2.circle(canvas, center, radius, (255, 0, 0), -1)

cv2.imshow("Circle Drawing", canvas)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

## Adding Custom Text (`cv2.putText`)

Used to label detected objects (e.g., printing "Face Detected" above a drawn rectangle).
- **Syntax:** `cv2.putText(image, text, origin, font, fontScale, color, thickness)`
- **Origin (`org`):** The Bottom-Left corner of where the text string will start.


```python
text = "OpenCV Drawing Master!"

# Bottom-left starting point for the text
org = (50, 480) 

# Font family (HERSHEY_SIMPLEX is standard and clean)
font = cv2.FONT_HERSHEY_SIMPLEX

# fontScale (1.0 is normal size)
font_scale = 1.0

# Color (White) and Thickness (2)
cv2.putText(canvas, text, org, font, font_scale, (255, 255, 255), 2)

cv2.imshow("Adding Text", canvas)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

## Common Mistakes
> **Warning**  
> - Forgetting that the Y-axis starts at the top and goes down. Coordinates are `(X, Y)`.
> - Mixing up BGR and RGB. If you want Red, you must use `(0, 0, 255)`, not `(255, 0, 0)`.
> - Providing float values for coordinates instead of integers. OpenCV drawing functions crash if given decimals.


## Key Takeaways
- Drawing permanently alters the pixel values of the image matrix in memory.
- `thickness = -1` fills in a shape completely.
- Text requires specifying a specific font family from the `cv2.FONT_HERSHEY_*` collection.
