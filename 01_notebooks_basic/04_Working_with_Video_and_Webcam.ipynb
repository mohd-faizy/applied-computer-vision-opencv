## The Core Concepts of Video

Before writing code, you need to understand the terminology:
- **Video:** A collection of images displayed very fast (usually 30 to 60 frames per second).
- **Frame:** A single, individual image within that video sequence. `1 Video = Many Frames`.
- **Frame-by-Frame Processing:** OpenCV pauses at every frame, applies logic, and moves to the next frame.


## Accessing the Webcam (`cv2.VideoCapture`)

To get a live feed from a camera, we use the `VideoCapture` function. We pass an integer (usually `0`) to tell OpenCV which camera to use.
- `0`: Default built-in webcam.
- `1` or `2`: External USB webcams.
- Or pass a string path like `"../data/videos/sample.mp4"` to read a saved video.


```python
# Note: Running infinite loops in Jupyter Notebooks can sometimes cause the kernel to freeze
# if not handled properly. Ensure the cv2.waitKey(1) block is written correctly.

# 1. Open the default camera
cap = cv2.VideoCapture(0)

# 2. Use a loop to continuously grab frames
while True:
    # ret: True if the frame was successfully read, False if it failed/ended
    # frame: The actual image matrix
    ret, frame = cap.read()
    
    if not ret:
        print("Error: Could not read frame.")
        break
        
    # 3. Display the live frame
    cv2.imshow("Webcam Feed", frame)
    
    # 4. Wait 1 millisecond for user input. If 'q' is pressed, break the loop.
    if cv2.waitKey(1) & 0xFF == ord('q'):
        print("Quitting...")
        break

# 5. Clean up! Release the camera so other apps can use it.
cap.release()
cv2.destroyAllWindows()
```

## Saving Video (`cv2.VideoWriter`)

Viewing a video is great, but saving it requires an encoder.
- **AVI Format (`.avi`):** Audio Video Interleave. A standard format.
- **FourCC (Codec):** A 4-character code used to compress frames. We use `XVID`.
- **FPS (Frames Per Second):** Determines how smooth the video is (e.g., `20.0`).


```python
camera = cv2.VideoCapture(0)

# Get the current Width and Height of the camera feed.
# OpenCV returns these as floats, so we MUST wrap them in int()
frame_width = int(camera.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(camera.get(cv2.CAP_PROP_FRAME_HEIGHT))

# Define the codec (compression format)
fourcc = cv2.VideoWriter_fourcc(*'XVID')

# Create the VideoWriter object: (Filename, Codec, FPS, (Width, Height))
recorder = cv2.VideoWriter('my_video.avi', fourcc, 20.0, (frame_width, frame_height))

while True:
    success, frame = camera.read()
    
    if not success:
        break
        
    # Write the current frame to the video file
    recorder.write(frame)
    
    # Show the user what is being recorded
    cv2.imshow("Recording Live", frame)
    
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Release both the camera AND the recorder object
camera.release()
recorder.release()
cv2.destroyAllWindows()
```

## Common Mistakes
> **Warning**  
> - Forgetting `cap.release()` can lock your webcam, forcing you to restart Python or your laptop.
> - Not matching the `frame_width` and `frame_height` in `VideoWriter` exactly to the dimensions of the incoming frame. If they mismatch, the video file will be 0 bytes and corrupted.
> - Using `cv2.waitKey(0)` instead of `cv2.waitKey(1)`. `0` waits indefinitely, which means your video will only update when you manually press a key!


## Key Takeaways

- `cv2.VideoWriter` encodes and saves sequences of images into a single video file.
- Everything relies on a continuous `while` loop grabbing frames using `.read()`.
