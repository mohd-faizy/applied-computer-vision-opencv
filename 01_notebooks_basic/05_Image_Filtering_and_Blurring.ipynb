## The Core Concepts of Filtering

- **Image Filtering:** Changing a single pixel's value based on the values of its neighbors.
- **Noise:** Unwanted random pixels (like dust or static).
- **The Kernel:** A tiny, square matrix (e.g., 3x3, 5x5) that slides across your entire image. It acts as the "instruction manual" for how the center pixel should change. **Kernels must always be odd numbers** so they have an exact center pixel.


## Load Sample Image
Load a sharp image to test blurring.


```
image_path = "../data/images/lena.jpg"
image = cv2.imread(image_path)

if image is None:
    print(f"Error: Image not found at {image_path}!")
else:
    print("Image loaded successfully.")
```

## Gaussian Blur (`cv2.GaussianBlur`)

Gaussian blur softens the image and removes harsh edges. It works by taking the **average** of the neighboring pixels.

**Syntax:** `cv2.GaussianBlur(source, (kernel_x, kernel_y), sigma)`
- **Kernel Size:** Must be an odd number like `(3,3)` or `(7,7)`. The higher the number, the stronger the blur.
- **Sigma:** Controls blur strength mathematically. `0` calculates it automatically.


```
if image is not None:
    # Apply Gaussian Blur with a 7x7 kernel
    blurred_image = cv2.GaussianBlur(image, (7, 7), 0)
    
    cv2.imshow("Original Sharp Image", image)
    cv2.imshow("Gaussian Blurred Image", blurred_image)
    
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

## Median Blur (`cv2.medianBlur`)

Unlike Gaussian, Median Blur sorts the neighboring pixels and picks the **exact middle value**.
- *Real-world use case:* Removing "Salt and Pepper" noise (random black and white dots).

**Syntax:** `cv2.medianBlur(source, kernel_size)`
- Notice you only pass a single integer for the kernel (e.g., `5`), not a tuple.


```
# For this, ideally load a noisy image. We will simulate noise or use the original.
if image is not None:
    # A kernel of 5 is strong enough to erase small dots
    clean_image = cv2.medianBlur(image, 5)
    
    cv2.imshow("Original Image", image)
    cv2.imshow("Cleaned Median Image", clean_image)
    
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

## Sharpening Filters (`cv2.filter2D`)

OpenCV doesn't have a built-in "Sharpen" function. Instead, we create our own custom math matrix (Kernel) using NumPy and apply it using `filter2D`.

**The Sharpening Logic:**
We create a 3x3 grid. We multiply the center pixel by a high number (`5`), and subtract (`-1`) the top, bottom, left, and right neighbors. This creates high contrast edges.


```
if image is not None:
    # Define the custom Sharpening Kernel using NumPy
    sharpen_kernel = np.array([
        [ 0, -1,  0],
        [-1,  5, -1],
        [ 0, -1,  0]
    ])
    
    # Apply the custom kernel to the image
    # -1 keeps the exact same color depth as the original image
    sharpened_image = cv2.filter2D(image, -1, sharpen_kernel)
    
    cv2.imshow("Original Image", image)
    cv2.imshow("Sharpened Crisp Image", sharpened_image)
    
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

## Common Mistakes
> **Warning**  
> - Passing an even number (like `4` or `(4,4)`) as a kernel size. It will crash because an even grid has no center pixel!
> - Over-blurring. Using a massive kernel (like `51x51`) takes a lot of processing power and completely destroys the image data.


## Key Takeaways
- Gaussian Blur = Average of neighbors (Good for general softening).
- Median Blur = Middle value of neighbors (Good for removing noise/dots).
- Custom Kernels can be applied using `cv2.filter2D` for effects like sharpening.
