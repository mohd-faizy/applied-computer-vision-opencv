## Canny Edge Detection (`cv2.Canny`)

Invented by John Canny in 1986, this algorithm detects edges by looking for areas where the pixel brightness changes suddenly and sharply.

> **Rule of Thumb**  
> Canny Edge Detection works best and fastest on **Grayscale** images.

**The Thresholds (T1 & T2):**
- **Threshold 1 (Lower Bound):** Detects weak/softer edges.
- **Threshold 2 (Upper Bound):** Detects strong/harsh edges.


```
image_path = "../data/images/sudoku.png"
# Load directly in Grayscale mode
gray_image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

if gray_image is not None:
    # Lower threshold = 50, Upper threshold = 150
    edges = cv2.Canny(gray_image, 50, 150)
    
    cv2.imshow("Original Grayscale", gray_image)
    cv2.imshow("Canny Edges (Outline)", edges)
    
    cv2.waitKey(0)
    cv2.destroyAllWindows()
else:
    print(f"Image not found at {image_path}")
```

## Image Thresholding (`cv2.threshold`)

Thresholding simplifies a grayscale image into a **Binary** image (strictly Black and White).
It acts as a strict "Yes or No" decision based on brightness.

**How it works:**
If the cut-off number is `120`:
- Brightness **> 120** becomes pure White (`255`).
- Brightness **< 120** becomes pure Black (`0`).


```
if gray_image is not None:
    # Syntax: cv2.threshold(source, threshold_value, max_value, method)
    # Returns two things: the threshold value used (ret), and the output image (thresh_img)
    ret, thresh_img = cv2.threshold(gray_image, 120, 255, cv2.THRESH_BINARY)
    
    cv2.imshow("Original Grayscale", gray_image)
    cv2.imshow("Binary Threshold Image", thresh_img)
    
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

## Bitwise Operations (AND, OR, NOT)

Bitwise operations apply logic gates directly to image pixels. They are used to mask, combine, or invert images.

- **AND (`cv2.bitwise_and`):** Keeps pixels where **both** inputs overlap. Used to *cut out* shapes.
- **OR (`cv2.bitwise_or`):** Keeps pixels if **either** input has white. Used to *merge* shapes.
- **NOT (`cv2.bitwise_not`):** Inverts colors (White -> Black, Black -> White).


```
# 1. Create two blank black canvases (300x300)
img1 = np.zeros((300, 300), dtype="uint8")
img2 = np.zeros((300, 300), dtype="uint8")

# 2. Draw a solid white circle on Canvas 1
cv2.circle(img1, (150, 150), 100, 255, -1)

# 3. Draw a solid white rectangle on Canvas 2
cv2.rectangle(img2, (100, 100), (250, 250), 255, -1)

# --- APPLYING BITWISE LOGIC ---

# AND: Only the overlapping section remains white
bitwise_and = cv2.bitwise_and(img1, img2)

# OR: Both shapes are merged
bitwise_or = cv2.bitwise_or(img1, img2)

# NOT: Inverts Image 1
bitwise_not = cv2.bitwise_not(img1)

cv2.imshow("Canvas 1 (Circle)", img1)
cv2.imshow("Canvas 2 (Rectangle)", img2)
cv2.imshow("Bitwise AND", bitwise_and)
cv2.imshow("Bitwise OR", bitwise_or)
cv2.imshow("Bitwise NOT", bitwise_not)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

## Common Mistakes
> **Warning**  
> - Forgetting to convert an image to Grayscale before applying thresholding.
> - Setting threshold values too low (everything becomes white) or too high (everything becomes black).
> - Trying to run bitwise operations on two images that have different dimensions (width and height must match exactly!).


## Key Takeaways
- `cv2.Canny` is best for detecting sharp outlines and borders.
- `cv2.threshold` creates a binary (B&W) image, which is the perfect setup for applying masks.
- `cv2.bitwise_and` is the industry standard way of applying a mask to cut an object out of a background.
